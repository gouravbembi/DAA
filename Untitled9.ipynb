```python
import cv2
import numpy as np

import cv2
from google.colab.patches import cv2_imshow
# Read input image
img = cv2.imread("/istockphoto-1149862124-612x612.jpg", cv2.IMREAD_UNCHANGED)

# Input border width
border_width = int(input("Enter border width (in pixels): "))

# Input border color
border_color = int(input("Enter border color (0-255): "))

# Add border
output = cv2.copyMakeBorder(
    img,
    border_width,
    border_width,
    border_width,
    border_width,
    cv2.BORDER_CONSTANT,
    value=border_color
)

# Display original image
print("Original Image:")
cv2_imshow(img)

# Display image with border
print("Image with Border:")
cv2_imshow(output)

# Save output
cv2.imwrite("output.png", output)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
img=cv2.imread("/istockphoto-1149862124-612x612.jpg",cv2.IMREAD_GRAYSCALE)
r=img.astype(float)
s=np.log(1+r)
s=np.uint8(s)
print("Original Image:")
cv2_imshow(img)
print("Log Transformed Image:")
cv2_imshow(s)
```
